```python
import numpy as np
```

```python
#set random seed (so results are the same every run)
np.random.seed(1234)
```

```python
#make a random 5x5 matrix with values between 10 and 100
X = np.random.randint(1, 51, (4, 4))
```

```python
#normalize the matrix (subtract mean, divide by std)
X_normalized = (X - X.mean()) / X.std()
```

```python
print("Matrix X:")
print(X)

print("\nNormalized X:")
print(X_normalized)

print("\nMean of normalized X:", X_normalized.mean())
print("STD of normalized X:", X_normalized.std())
```

B. Cubes Divisible by 4


```python
import numpy as np
```

```python
nums = np.arange(1, 101)
```

```python
cubes = nums ** 3
```

```python
C = cubes.reshape(10, 10)
```

```python
div_by_4 = C[C % 4 == 0]
```

```python
print("Shape of C:", C.shape)
```

```python
print("\nCubes divisible by 4:")
print(div_by_4)
```

```python
print("\nNumber of selected elements:", div_by_4.size)
```

```python
np.save("div_by_4.npy", div_by_4)
```

C. Above Mean Sqaures Problem

```python
import numpy as np
```

```python
nums = np.arange(1, 37)
```

```python
squares = nums ** 2
```

```python
S = squares.reshape(6, 6)
```

```python
S_mean = S.mean()
```

```python
above_mean = S[S > S_mean]
```

```python
print("Array S:")
print(S)
```

```python
print("\nMean of S:", S_mean)
```

```python
print("\nValues above the mean:")
print(above_mean)
```

```python
print("\nNumber of selected elements:", above_mean.size)
```

```python
np.save("above_mean.npy", above_mean)
```

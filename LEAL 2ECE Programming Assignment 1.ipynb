```python
def rotate_word(text):
    # if only one letter, just return it
    if len(text) == 1:
        return text
    # take everything except the first letter + add the first letter at the end
    return text[1:] + text[0]

# test the function
print(rotate_word("Anaconda"))  # adnocanA
print(rotate_word("Ultraviolet"))   # teloivartlU
print(rotate_word("Signals"))    # slangiS
print(rotate_word("Systems"))       # smetsyS
```

```python
def make_username(first_name, last_name):
    # turn everything lowercase
    first = first_name.lower()
    last = last_name.lower()
    
    # remove spaces
    first = first.replace(" ", "")
    last = last.replace(" ", "")
    
    # join with a period
    return first + "." + last

# test the function
print(make_username("Arven", "Leal"))        # arven.leal
print(make_username("Raiya", "Aestelle"))         # raiya.aestelle
print(make_username("Clarence Jasmine", "Gadacho"))   # clarencejasmine.gadacho
```

```python
def swap_bookends(items):
    # unpack the list
    first, *middle, last = items
    
    # make a new list with swapped ends
    return [last] + middle + [first]

# test the function
print(swap_bookends([10, 11, 12, 13, 14, 15]))      # [15, 14, 13, 12, 11, 10]
print(swap_bookends(["Digital", "Signals", "Communication"])) # ["Communication", "Signals", "Digital"]
print(swap_bookends([7, 6]))                   # [6, 7]
```
